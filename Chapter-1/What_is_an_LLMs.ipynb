# **What is an LLM?**

An LLM is a neural network designed to understand, generate, and respond to human- like text. These models are deep neural networks trained on massive amounts of text data, sometimes encompassing large portions of the entire publicly available text on the internet.

LLMs utilize an architecture called the transformer, which allows them to pay selec- tive attention to different parts of the input when making predictions, making them especially adept at handling the nuances and complexities of human language.

Since LLMs are capable of generating text, LLMs are also often referred to as a form of generative artificial intelligence, often abbreviated as generative AI or GenAI.

**The transformer architecture consists of two submodules**

1.**Encoder**

2.**Decoder**

The **encoder module** processes the input text and encodes it into a series of numericals representation or vectors that capture the contextual information of the input text.**Then**

The **decoder** takes these encoded vectrs and genrates the output.

**In a translation task For example**:

encoder would encode the text from the source language into vectors and decoder wuld decode these vectors to genrate text into target language.

Both encder and decoder consists of many connected layers by a so_called **Self_attention mechanism**.
